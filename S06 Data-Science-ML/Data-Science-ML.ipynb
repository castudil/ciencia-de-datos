¿Qué es Machine Learning?
1. Introducción
Motivación inicial: ¿Por qué hablar de ML en Ciencia de Datos?

Machine Learning podríamos decir que es una una disciplina que se encarga del estudio de algoritmos que son capaces de aprender de los datos. Hay tres elementos fundamentales cuando hablamos de aprendizaje automático: los datos, el algoritmo, y el modelo. 

Los datos son como la gasolina del aprendizaje automático. Son los ingredientes. Podríamos hacer una analogía con un queque que queremos cocinar.

La receta vendría ser el algoritmo que es la forma en que vamos a mezclar los datos. Finalmente el queque corresponde a la analogía del modelo, es decir es lo que obtenemos al procesar los datos a través del algoritmo de aprendizaje automático.

Este modelo tiene la capacidad de encontrar patrones en los datos que proporcionamos pero fundamentalmente es capaz de encontrar patrones en datos nuevos, no vistos hasta ahora. A esa característica se le llama generalización y es vital en esta área. De hecho es tan importante que cuando queramos medir la capacidad de un de un modelo, precisamente vamos a tratar de estudiar su capacidad de generalización. Esto lo hacemos de una manera bastante ingeniosa: vamos a separar los datos que tengamos a disposición en tres conjuntos. El primero le vamos a llamar conjunto de entrenamiento, también vamos a contar con un conjunto llamado de validación y un último conjunto llamado de Test o prueba. El conjunto de entrenamiento tiene por objetivo construir el modelo de aprendizaje a través del algoritmo. El conjunto de validación, que normalmente es un conjunto pequeño, lo vamos a utilizar para poder ajustar los parámetros del algoritmo de tal manera que el modelo quede optimizado. Finalmente el conjunto de prueba, que nunca participó en el proceso de entrenamiento, va a ser utilizado para verificar finalmente si el modelo de aprendizaje es capaz de reconocer patrones en entornos nuevos. 


Ejemplos cotidianos: recomendaciones en Netflix/Spotify, filtros de spam en correo, predicción del clima, diagnósticos médicos.
2. Definición básica
Definición simple: ML es un subcampo de la Inteligencia Artificial que permite a las computadoras aprender a partir de datos sin ser programadas explícitamente.
Diferencia con la programación tradicional:
Programación tradicional: reglas fijas → salida.
ML: datos → algoritmo aprende reglas → salida.
3. Elementos clave de ML
Datos: materia prima (ejemplo: registros de pacientes, imágenes, textos).
Modelo: representación matemática que aprende patrones.
Algoritmo de aprendizaje: el “recetario” que ajusta los parámetros del modelo.
Evaluación: medir qué tan bien funciona (accuracy, F1-score, etc.).
4. Tipos principales de aprendizaje
Supervisado
Entrenamiento con datos etiquetados (ej.: predecir precio de casas).
Problemas: clasificación y regresión.
No supervisado
Descubre patrones ocultos sin etiquetas (ej.: segmentación de clientes).
Problemas: clustering, reducción de dimensionalidad.
Por refuerzo
Aprender a través de ensayo-error, con recompensas (ej.: robots, juegos, sistemas de recomendación).
5. Flujo de trabajo en ML
Definición del problema.
Obtención y preparación de datos (limpieza, normalización, features).
Selección de modelo (lineal, árboles, redes, etc.).
Entrenamiento (ajuste de parámetros).
Evaluación y validación.
Despliegue en producción.
6. Ejemplos prácticos
Clasificar correos como spam / no spam.
Predecir la probabilidad de abandono de clientes.
Agrupar canciones según similitudes acústicas.
7. Limitaciones y desafíos
Necesidad de datos de calidad.
Sesgo en los datos → sesgo en el modelo.
Costo computacional.
Interpretabilidad: modelos complejos como “cajas negras”.
8. Conexión con Ciencia de Datos
ML es una herramienta dentro del ciclo de Ciencia de Datos.
Ciencia de Datos incluye: estadística, programación, visualización, comunicación de resultados.
ML aporta la capacidad de automatizar predicciones y descubrimiento de patrones.
9. Cierre
Reflexión: ML no reemplaza al científico de datos, sino que amplía sus capacidades.

El **muestreo estratificado** es una técnica de partición de datos en la que se divide el conjunto total en subgrupos homogéneos llamados *estratos* (por ejemplo, según la variable objetivo: sobrevivió/no sobrevivió). Luego, **se toma una muestra de cada estrato manteniendo la misma proporción que existe en el conjunto original**.

Esto es importante en el problema del Titanic porque la proporción de personas que sobrevivieron y no sobrevivieron está desbalanceada. Si se dividieran los datos de manera aleatoria, podríamos obtener conjuntos de entrenamiento o prueba con proporciones muy diferentes a las reales, lo que afectaría la capacidad del modelo para generalizar y evaluar correctamente su desempeño. El muestreo estratificado garantiza que tanto el conjunto de entrenamiento como el de prueba mantengan la misma distribución de la variable objetivo, permitiendo una evaluación más justa y representativa del modelo.

```python
from sklearn.svm import SVC
from xgboost import XGBClassifier

# Entrenar modelos
dt_model = DecisionTreeClassifier(random_state=42)
svm_model = SVC(random_state=42, probability=True)
xgb_model = XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)

dt_model.fit(X_train, y_train)
svm_model.fit(X_train, y_train)
xgb_model.fit(X_train, y_train)

# Predecir en el conjunto de validación
y_pred_dt = dt_model.predict(X_val)
y_pred_svm = svm_model.predict(X_val)
y_pred_xgb = xgb_model.predict(X_val)

# Calcular métricas
metrics = {
    'Modelo': [],
    'Accuracy': [],
    'Precision': [],
    'Recall': [],
    'F1-score': []
}

for name, y_pred in zip(['Decision Tree', 'SVM', 'XGBoost'], [y_pred_dt, y_pred_svm, y_pred_xgb]):
    metrics['Modelo'].append(name)
    metrics['Accuracy'].append(accuracy_score(y_val, y_pred))
    metrics['Precision'].append(precision_score(y_val, y_pred))
    metrics['Recall'].append(recall_score(y_val, y_pred))
    metrics['F1-score'].append(f1_score(y_val, y_pred))

# Crear tabla comparativa
metrics_df = pd.DataFrame(metrics)
print(metrics_df)
```

```python
from sklearn.preprocessing import StandardScaler

# Crear el escalador y ajustarlo solo con los datos de entrenamiento
scaler = StandardScaler()
X_train_z = scaler.fit_transform(X_train)

# Transformar los datos de validación y prueba usando el mismo escalador
X_val_z = scaler.transform(X_val)
X_test_z = scaler.transform(X_test) # Usar transform, no fit_transform

# Opcional: convertir de nuevo a DataFrame para mantener nombres de columnas
X_train_z = pd.DataFrame(X_train_z, columns=X_train.columns, index=X_train.index)
X_val_z = pd.DataFrame(X_val_z, columns=X_val.columns, index=X_val.index)
X_test_z = pd.DataFrame(X_test_z, columns=X_test.columns, index=X_test.index)
```

### ¿Por qué aplicar el escalado solo al grupo de entrenamiento?

Cuando aplicamos técnicas de escalado (como `StandardScaler`), el objetivo es transformar las variables numéricas para que tengan media 0 y desviación estándar 1. Sin embargo, **es fundamental que los parámetros de escalado (media y desviación estándar) se calculen únicamente usando el conjunto de entrenamiento**. Esto se debe a:

- **Simulación del mundo real:** En un escenario real, los datos de test representan información nueva y desconocida para el modelo. No deberíamos tener acceso a ellos durante el entrenamiento ni usar su información para ajustar ningún parámetro.
- **Evitar filtración de datos (data leakage):** Si calculamos la media y desviación estándar usando tanto entrenamiento como test, estamos introduciendo información del futuro (test) en el proceso de entrenamiento. Esto puede llevar a una sobreestimación del desempeño del modelo, ya que indirectamente el modelo “conoce” parte de los datos de test.
- **Generalización:** El objetivo es que el modelo aprenda a partir de los datos de entrenamiento y luego se evalúe su capacidad de generalizar a datos nuevos. Si usamos información del test para el escalado, la evaluación deja de ser honesta.

**En resumen:**  
- Calcula la media y desviación estándar solo con el entrenamiento (`fit`).
- Aplica la transformación a entrenamiento, validación y test (`transform`).
- Así, garantizas que el modelo no tiene acceso a información del test durante el entrenamiento y la evaluación será realista.

```python
# Entrenar modelos con los datos escalados
dt_model_z = DecisionTreeClassifier(random_state=42)
svm_model_z = SVC(random_state=42, probability=True)
xgb_model_z = XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)

dt_model_z.fit(X_train_z, y_train)
svm_model_z.fit(X_train_z, y_train)
xgb_model_z.fit(X_train_z, y_train)

# Predecir en el conjunto de validación escalado
y_pred_dt_z = dt_model_z.predict(X_val_z)
y_pred_svm_z = svm_model_z.predict(X_val_z)
y_pred_xgb_z = xgb_model_z.predict(X_val_z)

# Calcular métricas
metrics_z = {
    'Modelo': [],
    'Accuracy': [],
    'Precision': [],
    'Recall': [],
    'F1-score': []
}

for name, y_pred in zip(['Decision Tree (scaled)', 'SVM (scaled)', 'XGBoost (scaled)'],
                        [y_pred_dt_z, y_pred_svm_z, y_pred_xgb_z]):
    metrics_z['Modelo'].append(name)
    metrics_z['Accuracy'].append(accuracy_score(y_val, y_pred))
    metrics_z['Precision'].append(precision_score(y_val, y_pred))
    metrics_z['Recall'].append(recall_score(y_val, y_pred))
    metrics_z['F1-score'].append(f1_score(y_val, y_pred))

# Crear tabla comparativa
metrics_df_z = pd.DataFrame(metrics_z)
print(metrics_df_z)
```

```python
# Clasificador manual basado en las reglas del árbol (variable 'rules') usando solo if/else.
# Opera sobre el dataframe numérico codificado como X_train_num / X_test_num (columnas: Pclass, Sex, SibSp, Parch, Fare, Embarked)

def _rule_classifier_row(row):
    # Variables (ya codificadas): Sex (0=female,1=male), Embarked (0=C,1=Q,2=S)
    if row['Sex'] <= 0.5:  # female
        if row['Pclass'] <= 2.5:
            if row['Fare'] <= 29.36:
                if row['Fare'] <= 28.23:
                    return 1
                else:
                    return 0
            else:
                return 1  # Ambas hojas dan 1
        else:  # Pclass > 2.5
            if row['Fare'] <= 23.35:
                if row['Embarked'] <= 1.5:  # C o Q
                    return 1
                else:  # S
                    return 0
            else:
                if row['Parch'] <= 0.5:
                    return 1
                else:
                    return 0
    else:  # male
        if row['Fare'] <= 15.17:
            return 0  # Todas las hojas de ese subárbol son 0
        else:
            if row['SibSp'] <= 2.5:
                if row['Fare'] <= 15.37:
                    return 1
                else:
                    return 0
            else:
                return 0

def rule_based_predict(df_num):
    return df_num.apply(_rule_classifier_row, axis=1)

# Aplicar al conjunto de test numérico ya existente (X_test_num) y evaluar
y_pred_ifelse = rule_based_predict(X_test_num)
acc_ifelse = accuracy_score(y_test_tree, y_pred_ifelse)

print("Accuracy clasificador if/else (reglas árbol):", acc_ifelse)

# (Opcional) matriz de confusión
cm = confusion_matrix(y_test_tree, y_pred_ifelse)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=[0,1])
disp.plot(values_format='d')
plt.title("Matriz de confusión - Clasificador if/else")
plt.show()

# Mostrar primeras predicciones
print("Primeras predicciones vs real:")
print(
    pd.DataFrame({
        'Predicción': y_pred_ifelse.head(10).values,
        'Real': y_test_tree.head(10).values
    })
)
```

Un clasificador construido solo con sentencias if/else (derivado de las reglas de un árbol de decisión) demuestra que detrás de una aparente lógica “hecha a mano” existe todo el proceso formal de Machine Learning: preparación de datos, selección de variables, particiones estratificadas, prevención de leakage, ajuste del modelo y validación. La extracción de reglas ofrece interpretabilidad, reproducibilidad y facilidad de despliegue (no dependes de librerías en producción), pero también expone limitaciones: 1) la lógica queda “congelada” y no se adapta a nuevos patrones si cambian los datos, 2) las reglas reflejan sesgos y decisiones del set de entrenamiento, 3) se pierde calibración probabilística y capacidad de combinar señales suaves, 4) escalar o mejorar requiere re-entrenar y volver a traducir, 5) es frágil ante pequeñas variaciones (umbrales duros). En síntesis, el if/else no sustituye al enfoque de ML; es una representación estática, útil para explicar, auditar y eventualmente integrar en sistemas legacy, pero su valor proviene precisamente de haber sido inducido con metodología rigurosa de ciencia de datos. La práctica madura consiste en: entrenar modelos, extraer reglas cuando se necesita interpretabilidad operacional, monitorear drift y re-generar las reglas según métricas objetivas (accuracy, recall, F1) y criterios de negocio. Así se combina la transparencia de la lógica explícita con la solidez estadística del proceso de aprendizaje.
